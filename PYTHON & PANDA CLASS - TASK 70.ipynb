```python
# for loops handling dictionaries

mycountry_dict = {'Name': 'Grey', 'Age':30, 'Favs': ('Vanilla', 'Strawberry')}
```

```python
for president in mycountry_dict.values():
    print (president)
```

```python
# looking through the values of dictionary

for president in mycountry_dict.values():
    print (president + president)
```

```python
# looking through the values and keys of dictionary

for key, value in mycountry_dict.items():
    print (key, '->', value)
```
